```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('C:/Users/user/Downloads/boston_housing.csv')
```

```python
# 1. 각 column의 평균 값을 bar plot으로 시각화
column_means = data.mean(numeric_only=True)
```

```python
plt.figure(figsize=(6, 3))
column_means.plot(kind='bar', color='skyblue')
plt.title('각 열의 평균 값')
plt.ylabel('Mean Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# 각 열의 평균값을 바 차트로 시각화한 결과, 평균값의 스케일 차이가 크다는 것을 알 수 있다.
# 예를 들어, 재산세율(TAX)나 범죄율(CRIM)은 다른 변수에 비해 훨씬 큰 값의 평균을 가지고 있으며,
# Price(주택 가격)의 평균은 약 22~23 수준이다.
```

```python
# 2. 각 열의 관계성 시각화 (히트맵)
plt.figure(figsize=(6,3))
correlation_matrix = data.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('전체 변수 간의 상관관계 히트맵')
plt.show()
# 상관계수 히트맵을 통해 각 열 간의 관계를 파악할 수 있다.
# 예를 들어 RM(방 수)과 Price는 양의 상관관계(0.7 이상)를 가지며,
# LSTAT(하위계층 비율)은 Price와 강한 음의 상관(-0.74)을 보인다.
# 이는 RM이 많을수록 가격이 비싸고, 하위계층 비율이 높을수록 가격이 낮다는 것을 의미한다.
```

```python
# 3. Price 열과 상관관계가 높은 2개의 열 찾기
price_corr = correlation_matrix['Price'].drop('Price')
top2_features = price_corr.abs().sort_values(ascending=False).head(2)
top2_names = top2_features.index.tolist()
```

```python
# 상관계수 히트맵
plt.figure(figsize=(5, 4))
sns.heatmap(data[['Price'] + top2_names].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Price와 상관관계 높은 변수 히트맵')
plt.show()
# Price와 가장 상관관계가 높은 변수는 RM(양의 상관)과 LSTAT(음의 상관)이다.
# RM은 집의 크기와 연관 있고, LSTAT은 저소득층 비율로 해석되므로
# 이 두 변수는 주택 가격을 예측하는 중요한 특징으로 활용 가능하다.
```

```python
# 4. CHAS에 따라 색 구분한 scatter plot (예시: RM vs LSTAT)
plt.figure(figsize=(4,3 ))
sns.scatterplot(data=data, x='RM', y='LSTAT', hue='CHAS', palette='Set1')
plt.title('CHAS 값에 따른 RM vs LSTAT 산점도')
plt.xlabel('방의 평균 개수 (RM)')
plt.ylabel('하위계층 비율 (LSTAT)')
plt.legend(title='CHAS')
plt.tight_layout()
plt.show()
# 산점도를 통해 CHAS가 0 또는 1일 때 RM과 LSTAT의 분포 차이를 시각적으로 확인할 수 있다.
# 대체로 CHAS가 1(강 주변)의 데이터는 더 고급 주택이 많으며,
# 이는 RM이 높고 LSTAT이 낮은 위치에 집중되어 있는 것을 통해 알 수 있다.
```

```python
# 분석 내용 
- 보스턴 주택 데이터셋은 다양한 변수들로 주택 가격(Price)을 예측하는 데 활용된다.
- 상관계수를 보면, 방의 수(RM), 범죄율(CRIM), 하위계층 비율(LSTAT), NOX 농도 등이
  Price와 밀접한 관련이 있음을 확인할 수 있다.
- 특히, LSTAT은 하위계층 비율을 나타내며 주택 가격과 가장 강한 음의 상관관계를 보여,
  이 지표가 높을수록 주택 가격은 낮아지는 경향을 보인다.
- RM은 가격과 강한 양의 상관관계를 보여, 넓은 집일수록 가격이 높은 일반적 패턴을 보여준다.
- CHAS는 찰스강과의 인접 여부를 나타내며, 강 인근은 일반적으로 고가 주택 분포가 많을 가능성이 있다.
```
