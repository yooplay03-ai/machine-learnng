```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic= sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.info()
```

```python
sns.catplot(x = 'sex', y='survived', data=titanic, kind= 'bar',
            height= 2.5, aspect= 1.5)
```

```python
sns.barplot( x= 'sex', y='survived', data= titanic) 
```

```python
titanic['class'].unique()
```

```python
sns.catplot(x='sex', y='survived', data=titanic, kind='bar',
            hue='class', height=2.5, aspect=2, palette='Set3')
```

```python
sns.catplot(x='sex', y='survived', data=titanic, kind='point',
            hue='class', height=2.5, aspect=2, palette='Set1') 
```

```python
sns.catplot(x = 'class', data= titanic, kind= 'count',
            height= 2.5, aspect= 1.5)
```

```python
sns.catplot(x='class', data=titanic, kind='count',
            hue='sex', height=2.5, aspect=1.5)
```

```python
sns.catplot(x='class', data=titanic, kind='count',
            hue='sex', height=2.5, aspect=1.5, dodge= False) 
```

```python
sns.barplot(x = 'sex', y='survived', data=titanic)
```
