```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
tips.info()
```

```python
sns.catplot(x='day',y='total_bill',data=tips,
            height=2.5, aspect=1.5)
```

```python
sns.catplot(y='day',x='total_bill',data=tips,
            height=2.5, aspect=1.5)
```

```python
sns.catplot(x='day',y='total_bill',data=tips,
            height=2.5, aspect=1.5, order=['Fri','Sat','Sun']) 
```

```python
sns.catplot(x='day',y='total_bill',data=tips,
            height=2.5, aspect=1.5, jitter=False) 
```

```python
sns.catplot(x='day',y='total_bill',data=tips,
            height=2.5, aspect=1.5, kind='swarm') 
```

```python
sns.swarmplot(x='day',y='total_bill',data=tips)
```

```python
sns.catplot(x='day',y='total_bill', data=tips,
            kind='swarm',height=3, aspect=1.5, hue='size')
```

```python
sns.swarmplot(x='day',y='total_bill',data=tips, hue='size')
```

```python
sns.catplot(x='day', y='total_bill', data=tips,
            height=2.5, aspect=1.5, kind='box')
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
```

```python
sns.catplot(x='day',y='total_bill', data=tips,
            height=2.5, aspect=1.5, kind='box', palette='Set3', hue='day')
```

```python
sns.catplot(x='day', y='total_bill', data=tips,
            height=2.5, aspect=1.5, kind='box', palette='Set1',
           hue='sex') 
```

```python
sns.catplot(x='day', y='total_bill', data=tips,
            height=2.5, aspect=1.5, kind='box', palette='Set2',
           hue='sex', col='size') 
```

```python
sns.catplot(x='day', y='total_bill', data=tips,
            height=2.5, aspect=1.5, kind='box', palette='Set1',
           hue='sex', dodge=False) 
```
