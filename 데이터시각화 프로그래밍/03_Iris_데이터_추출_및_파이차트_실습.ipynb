```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset('iris')
iris = iris.drop('species', axis = 1)
iris.head()
```

```python
iris1 = iris.iloc[0]
iris1
```

```python
iris1.plot.pie()
```

```python
iris2 = iris.iloc[0:3]
```

```python
iris2.plot.pie(subplots = True)
```
