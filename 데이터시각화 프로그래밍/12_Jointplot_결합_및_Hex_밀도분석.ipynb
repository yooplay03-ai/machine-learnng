```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips,
              height=4)
```

```python
sns.jointplot( x= 'total_bill', y='tip', data=tips,
              kind= 'hex', height=4)
```

```python
sns.jointplot( x= 'total_bill', y='tip', data=tips,
              kind= 'kde', height=4) 
```

```python
sns.jointplot( x= 'total_bill', y='tip', data=tips,
              kind= 'kde',fill=True, height=4, cmap= 'Oranges')  
```

```python
sns.pairplot(tips, height= 1.5)
```

```python
g = sns.PairGrid(tips, height=1.5)
g.map(sns.scatterplot)
```

```python
g = sns.PairGrid(tips, height=1.5)
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(tips, height=1.5)
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.kdeplot, fill= True, cmap = 'Greens', n_levels=5)
```

```python
sns.regplot(x='total_bill', y='tip', data=tips, ci=70)
```

```python
sns.regplot(x='total_bill', y='tip', data=tips, ci=70) 
```
