```python
import pandas as pd
import numpy as np
import seaborn as sns
import plotly as py
from plotly.offline import iplot
import cufflinks as cf
```

```python
py.offline.init_notebook_mode(connected = True)
cf.go_offline()
```

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic.iplot(x = 'sex', y = 'survived', kind = 'bar')
```

```python
iris = sns.load_dataset('iris')
iris = iris.drop('species', axis = 1)
iris
```

```python
iris.iplot(kind= 'bar')
```

```python
iris.iplot(kind = 'bar', barmode ='stack', bargap=0.7)
```

```python
iris.iplot(kind = 'barh', barmode ='stack', bargap=0.7) 
```

```python
iris.iplot(kind='box')
```

```python
iris.iplot(kind='area', fill='True')
```

```python
iris.iplot(fill=True)
```

```python
iris.iplot(kind='surface')
```

```python
iris.iplot(kind='hist', bargap=0.5)
```

```python
iris.corr()
```

```python
iris.corr().iplot(kind='heatmap',colorscale='PiYG')
```
