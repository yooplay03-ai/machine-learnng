```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset('iris')
iris = iris.drop('species', axis = 1)
iris
```

```python
plt.boxplot(iris, labels = iris.columns)
plt.show()
```

```python
plt.boxplot(iris, labels=iris.columns, vert = False)
plt.show()
```

```python
plt.boxplot(iris, labels=iris.columns, patch_artist = True,
            boxprops = {'linewidth':3, 'color':'blue','facecolor':'magenta'}, 
            medianprops = {'linewidth':2, 'linestyle':'--','color':'yellow'},
            whiskerprops = {'linewidth':2, 'linestyle':'-','color':'red'},
            capprops = {'linewidth':3, 'color':'green'},
            flierprops = {'marker':'d','markerfacecolor':'cyan','markersize':7,'alpha':0.5,'markeredgecolor':'none'}
           )
plt.title('아이리스 박스 플롯')
plt.show()
```

```python
x = np.linspace(-2,2,20)
x
```

```python
y1 = 0.5 * (x - 1)**2-1
y2 = -0.7 * (x * 0.5)**2 + 0.5
```

```python
y2
```

```python
plt.plot(x,y1)
plt.plot(x,y2)
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.subplot(1,2,2,)
plt.plot(x,y2)
```

```python
plt.figure(figsize = (10,4))

plt.subplot(2,1,1)
plt.plot(x,y1,marker = 'o', color = 'red', markersize = 8)

plt.subplot(2,1,2)
plt.plot(x,y2,marker = '^', color = 'blue', markersize = 8, linestyle = '--')
```

```python
fig = plt.figure(figsize = (4,3))

ax1 = fig.add_axes([0,0,1,1])
ax2 = fig.add_axes([0.12,0.6,0.4,0.3])

ax1.plot(x,y1,color = 'red')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('Y1 plot')

ax2.plot(x,y1,color = 'blue')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Y2 plot')
```
