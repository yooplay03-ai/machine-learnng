```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2,2,20)
x
```

```python
y = 0.5 * (x - 1)**2-3
y
```

```python
plt.plot(y)
```

```python
plt.plot(x, y)
```

```python
plt.plot(x,y,color = 'purple', marker = 'o', markersize = 10)
```

```python
plt.plot(x,y,color = 'purple', marker = 'o', markersize = 7, linestyle = '-.')
```

```python
plt.plot(x,y,color = 'purple', marker = 'o', markersize = 7, linestyle = '-.')
plt.title('Line plot')
```

```python
plt.plot(x,y,color = 'purple', marker = 'o', markersize = 7, linestyle = '-.')
plt.title('Line plot')
plt.xlabel('x axis')
plt.ylabel('y axis')
```

```python
plt.plot(x,y,color = 'purple', marker = 'o', markersize = 7, linestyle = '-.')
plt.title('Line plot', fontsize = 20, color = 'red')
plt.xlabel('x axis', fontsize = 15, color = 'green')
plt.ylabel('y axis', fontsize = 15, color = 'green')
plt.grid(linestyle = '--', color = 'pink')
```

```python
iris = sns.load_dataset('iris')
iris = iris.drop('species', axis = 1)
iris.head()
```

```python
x = iris['petal_length']
y = iris['petal_width']
```

```python
plt.scatter(x, y, linewidth = 5)
```

```python
plt.scatter(x,y, linewidth = 5, color = 'green', alpha = 0.5)
plt.title('Scatter plot')
plt.xlabel('petal length')
plt.ylabel('petal width')
plt.grid()
```

```python
iris_mean = iris.mean()
iris_mean
```

```python
x = iris_mean.index
y = iris_mean
plt.bar(x, y)
```

```python
plt.bar(x, y, width = 0.7, color = 'purple', alpha = 0.5)
```

```python
data = iris['sepal_length']
data
```

```python
plt.hist(data, bins = 7, rwidth = 0.7, color = 'cyan', alpha = 0.3)
plt.title('Histogram', fontsize = 20, color = 'red')
plt.xlabel('Sepal length', fontsize = 18, color = 'blue')
plt.grid()
```
