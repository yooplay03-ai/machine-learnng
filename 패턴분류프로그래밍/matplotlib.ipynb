```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y1 = [5,4,3,2,1]
y2 = [1,2,3,4,5]
plt.plot(x,y1,color='red', marker='o',markersize=10, linestyle=':',linewidth=3, label='line1')
plt.plot(x,y2,color='green',marker='s',markersize=10, linestyle='--',linewidth=3, label ='line2')
plt.legend()
plt.title('line plot')
plt.xlabel('X')
plt.ylabel('y')
plt.grid()
```

```python
x1 = [2,3,1,7,10,5,3]
y1 = [7,2,9,1,3,1,11]

x2 = [-2,-3,-1,-7,-10,-5,-3]
y2 = [-7,-2,-9,-1,-3,-1,-11]

plt.scatter(x1,y1,s=100,marker='^', color = 'magenta')
plt.scatter(x2,y2,s=100,marker='d')
```
