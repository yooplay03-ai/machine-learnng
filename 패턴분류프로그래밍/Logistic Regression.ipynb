```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('C:/Users/user/Downloads/classification data/Social_Network_Ads.csv').values
dataset
```

```python
X = dataset[:,:-1]
y = dataset[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25, random_state=42)
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
sc.fit(X_train)
Xsc_train = sc.transform(X_train)
Xsc_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test,y_pred)
print(acc)
```

```python
w = lr.coef_
b = lr.intercept_
print(w)
print(b)
```

```python
X1 = np.array([[50,50000]])
X1_sc = sc.transform(X1)
y1_pred = lr.predict_proba(X1_sc)
print(y1_pred)
```
