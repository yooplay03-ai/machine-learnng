```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,11,1)
y = np.arange(-10,-5,1)
print(x)
print(y)
```

```python
x = np.linspace(1,10,10)
y = np.linspace(-10,-6,5)
print(x)
print(y)
```

```python
X,Y = np.meshgrid(x,y)
print(X)
print(Y)
```

```python
plt.scatter(X,Y)
```

```python
x = np.linspace(1,5,5)
y = np.linspace(1,5,5)
X,Y = np.meshgrid(x,y)
Z = np.array([[10,10,10,10,10],
             [10,5,5,5,10],
             [10,5,1,5,10],
             [10,5,5,5,10],
             [10,10,10,10,10]])
plt.contourf(X,Y,Z,cmap= 'Reds')
plt.colorbar()
plt.scatter(X,Y)
```
